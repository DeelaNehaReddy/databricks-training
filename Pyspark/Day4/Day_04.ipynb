```python
df=spark.read.format("csv").option('header',True).option('inferSchema',True).load("/Volumes/workspace/default/databricks/Big Sales.csv")
```

### LIMIT

```python
df.limit(10).display()
```

### DROP

```python
df.drop("Item_Weight").display()
```

```python
##Drop two columns
df.drop("Item_Visibility","Item_Type").display()
```

### DROP DUPLICATES

```python
# Drop rows with same values
df.dropDuplicates().display()
```

```python
#Drop column which have duplicate values
df.dropDuplicates(["Item_Type","Item_Fat_Content"]).display()
```

### CREATING DATA FRAMES

```python
data = [
    (1, "a"),
    (2, "b"),
    (3, "c")
]
schema = ["id", "value"]
df1 = spark.createDataFrame(data, schema)
df1.display()
```

```python
data2 = [
    ("a1",10),
    ("b1", 20)
]
schema2 = [ "value","id"]
df2 = spark.createDataFrame(data2, schema2)
df2.display()
```

### UNION

```python
df1.union(df2).display()
```

### UNIOUN BY NAME

```python
df1.unionByName(df2).display()
```

### STRING FUNCTIONS

```python
#Upper Case
from pyspark.sql.functions import *
df.select(upper("Item_Type").alias("Item_Type_Upper")).display()
```

```python
#Lower Case
df.select(lower("Item_Type").alias("Item_Type_Lower")).display()
```

```python
#initcap
df.select(initcap("Item_Type").alias("Item_Type_initcap")).display()
```
