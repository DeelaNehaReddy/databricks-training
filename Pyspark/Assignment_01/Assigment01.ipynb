###WITHCOLUMN

```python
#Create bonus column as 10% of salary.
df.withColumn("bonus",col("salary")*0.10).display()
#Create annual_salary column.
df.withColumn("annual_salary",col("salary")*12).display()
#Create tax column as 5% of salary.
df.withColumn("updated_salary",col("salary")+5000).display()
#Create salary_category column:High,Medium,Low
from pyspark.sql.functions import col, when
df.withColumn("salary_category",when(col("salary")>60000,"High").when(col("salary")>40000,"Medium").otherwise("Low")).display()
```

```python
#Create age_group column:Young,Adult
from pyspark.sql.functions import *
df.withColumn("age_group",when(col("age")<18,"Young").when(col("age")>=18,"Adult")).display()
#Create location column by combining city and department.
df.withColumn("location",concat(col("city"),lit('- '),col("department"))).display()
#Create increment_salary column with 15% hike.
df.withColumn("increment_salary",col("salary")*1.15).display()
#Create name_length column using employee name.
df.withColumn("name_length",length(col("emp_name"))).display()
#Create is_high_salary column using condition.
df.withColumn("is_high_salary",when(col("salary")>60000,"High").otherwise("Low")).display
#Create joining_year column from joining_date.
from pyspark.sql.functions import col, year
df = df.withColumn(
    "joining_date",
    to_date(col("joining_date"), "yyyy-MM-dd")
)
df.withColumn("joining_year",year(col("joining_date"))).display()
#Create salary_after_tax column.
df.withColumn("salary_after_tax",col("salary")*0.95).display()
#Create department_code column.
df.withColumn("department_code",concat(col("department"),lit("-"),col("emp_id"))).display()
#Create double_salary column.
df.withColumn("double_salary",col("salary")*2).display()
```

###LIMIT

```python
#Display first 5 records.
df.limit(5).display()
#Display top 10 employees.
df.limit(10).display()
#Display first 3 employees from IT department.
df.filter(col("department")=="IT").limit(3).display()
#Display top 5 highest salary employees.
df.orderBy(col("salary").desc()).limit(5).display()
#Display lowest 5 salary employees.
df.orderBy(col("salary").asc()).limit(5).display()
#Display first 7 rows after sorting by age.
df.orderBy(col("age").asc()).limit(7).display()
#Display first 2 employees from Hyderabad.
df.filter(col("city")=="Hyderabad").limit(2).display()
#Display first 15 records from dataframe.
df.limit(15).display()
#Display top 5 youngest employees.
df.orderBy(col("age").asc()).limit(5).display()
#Display first 8 employees after filtering salary > 60000
df.filter(col("salary")>60000).limit(8).display()
```

###SORT / ORDERBY

```python
#Sort employees by salary ascending.
df.orderBy(col("salary").asc()).display()
#Sort employees by salary descending.
df.orderBy(col("salary").desc()).display()
#Sort employees by age descending.
df.orderBy(col("age").desc()).display()
#Sort employees by emp_name ascending.
df.orderBy(col("emp_name")).display()
#Sort employees by city and salary descending.
df.orderBy(col("city").desc(), col("salary").desc()).display()
#Sort employees by joining_date.
df.orderBy(col("joining_date")).display()
#Sort employees by department.
df.orderBy(col("department")).display()
#Sort employees by designation descending.
df.orderBy(col("designation").desc()).display()
#Sort employees first by city then age.
df.orderBy(col("city"),col("age")).display()
#Sort employees by salary and limit top 10.
df.orderBy(col("salary")).limit(10).display()
#Sort employees by emp_id descending.
df.orderBy(col("emp_id").desc()).display()
#Sort employees from IT department by salary descending.
df.filter(col("department")=="IT").orderBy(col("salary").desc()).display()
#Sort employees by joining_date descending.
df.orderBy(col("joining_date").desc()).display()
#Sort employees alphabetically by emp_name.
df.orderBy(col("emp_name")).display()
#Sort employees by multiple columns.
df.orderBy(col("department"), col("salary").desc(), col("age")).display()
```

###TYPECASTING

```python
#Convert salary to string datatype.
df.withColumn("salary",col("salary").cast("string")).display()
#Convert age to double datatype.
df.withColumn("age",col("age").cast("double")).display()
#Convert joining_date to date datatype.
df.withColumn("joining_date",col("joining_date").cast("date")).display()
#Convert emp_id to string datatype.
df.withColumn("emp_id",col("emp_id").cast("string")).display()
#Convert salary to integer datatype.
df.withColumn("salary",col("salary").cast("int")).display()
#Convert age to string datatype.
df.withColumn("age",col("age").cast("string")).display()
#Convert joining_date to timestamp datatype.
df.withColumn("joining_date",col("joining_date").cast("timestamp")).display()
#Convert salary to float datatype.
df.withColumn("salary",col("salary").cast("float")).display()
#Convert emp_id to long datatype.
df.withColumn("emp_id",col("emp_id").cast("long")).display()
#Convert multiple columns into different datatypes.
df.withColumn("salary", col("salary").cast("float")) \
  .withColumn("age", col("age").cast("string")) \
  .withColumn("joining_date", col("joining_date").cast("date")) \
  .display()
```
